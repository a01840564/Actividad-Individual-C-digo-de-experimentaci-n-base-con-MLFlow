```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

DATA_PATH = Path.cwd() / "train.csv"
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (8, 5)
print(f"Dataset: {DATA_PATH}")
```

## Paso 1 — Carga y descripción del dataset


```python
if not DATA_PATH.exists():
    raise FileNotFoundError(
        f"No encontré {DATA_PATH}. Descarga train.csv de Kaggle y ponlo en data/."
    )

df_raw = pd.read_csv(DATA_PATH)
df = df_raw.copy()

print("Tamaño del dataset (filas, columnas):", df.shape)
print("\nTipos de cada columna:")
print(df.dtypes)
print("\nPrimeras filas:")
display(df.head())

print("\nCuántos nulos hay por columna:")
display(df.isna().sum().to_frame("nulos"))

# Significado breve de las columnas
VARIABLES = {
    "PassengerId": "ID del pasajero (solo identificador, no lo uso para predecir).",
    "Survived": "Lo que quiero predecir: 0 = murió, 1 = sobrevivió.",
    "Pclass": "Clase del boleto (1, 2 o 3).",
    "Name": "Nombre (no lo uso en el modelo base).",
    "Sex": "Sexo (male / female).",
    "Age": "Edad.",
    "SibSp": "Hermanos o esposo/a a bordo.",
    "Parch": "Padres o hijos a bordo.",
    "Ticket": "Número de boleto (no lo uso).",
    "Fare": "Precio del boleto.",
    "Cabin": "Cabina (casi siempre viene vacía).",
    "Embarked": "Puerto donde subió (C, Q o S).",
}

pd.DataFrame(
    {"variable": list(VARIABLES.keys()), "descripcion": list(VARIABLES.values())}
)
```

## Paso 2 — Análisis exploratorio (EDA)

En esta sección se revisan estadísticas y algunos gráficos para ver cómo se distribuyen las variables y si hay patrones con la supervivencia


```python
print("Estadísticas numéricas:")
display(df.describe())

print("Conteo de la variable objetivo:")
display(df["Survived"].value_counts().to_frame("conteo"))

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.countplot(data=df, x="Survived", ax=axes[0])
axes[0].set_title("Distribución de Survived")

sns.countplot(data=df, x="Sex", hue="Survived", ax=axes[1])
axes[1].set_title("Supervivencia por sexo")
plt.tight_layout()
plt.show()

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(df["Age"].dropna(), kde=True, ax=axes[0])
axes[0].set_title("Distribución de Age")

sns.boxplot(data=df, x="Survived", y="Fare", ax=axes[1])
axes[1].set_title("Fare vs Survived")
plt.tight_layout()
plt.show()

numeric_cols = ["Age", "SibSp", "Parch", "Fare"]
corr = df[numeric_cols + ["Survived"]].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Matriz de correlación (variables numéricas)")
plt.show()
```
